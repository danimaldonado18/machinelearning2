# Regresión Lineal

```python
#Importamos las librerias
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt 
```

```python
#Creamos el conjunto de datos
df=pd.DataFrame({
    "horas":[1,2,3,4,5],
    "notas":[2,3,5,4,6]
})
X=df[["horas"]]
Y=df[["notas"]]
print(X)
```

```python
#Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X,Y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict(X)
```

```python
#Realizamos la nueva prediccion 7
horas = float(input("Ingrese las horas de estudio: "))
prediccion = modelo.predict([[horas]])
print("Prediccion: ", prediccion)
```

```python
plt.figure(figsize=(8,4))

# 1. Graficamos los datos reales en azul
plt.scatter(X, Y, color='blue', label='Datos reales', s=50)

# 2. Trazamos la línea de regresión en rojo
plt.plot(X, modelo.predict(X), color='red', label='Línea de regresión')

# 3. Graficamos las predicciones del modelo para los datos originales en naranja
plt.scatter(X, modelo.predict(X), color='orange', label='Predicciones del modelo', s=50)

# 4. Graficamos tu dato predicho individual en verde (más grande)
plt.scatter(horas, prediccion, color='green', s=150, label='Predicción')

# Configuramos las etiquetas
plt.xlabel('Horas de estudio')
plt.ylabel('Nota')
plt.title('Regresión Lineal: Datos Reales vs Predicciones')
plt.legend()

plt.show()
```
